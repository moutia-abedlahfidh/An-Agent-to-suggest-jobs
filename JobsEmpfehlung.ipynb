```python
from PyPDF2 import PdfReader 
```

```python
myCV = "Lebenslauf Moutia Abed Lahfidh.pdf"
```

```python
reader = PdfReader(myCV)
```

```python
def getWords(reader):
    for page in reader.pages:
        text = page.extract_text()
        print(text)
```

```python
print(getWords(reader))
```
